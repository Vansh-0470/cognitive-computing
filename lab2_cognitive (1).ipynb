```python
# Q3. Random Numbers

import random
from collections import Counter

# Enter roll number
roll_no = input("Enter your roll number: ")

# Set random seed using roll number
random.seed(int(roll_no))

# Generate 100 random numbers between 100 and 900
numbers = [random.randint(100, 900) for _ in range(100)]

print("Random numbers:")
print(numbers)

# Odd numbers
odd_numbers = [x for x in numbers if x % 2 != 0]

print("\nOdd numbers:")
print(odd_numbers)
print("Count of odd numbers:", len(odd_numbers))

# Even numbers
even_numbers = [x for x in numbers if x % 2 == 0]

print("\nEven numbers:")
print(even_numbers)
print("Count of even numbers:", len(even_numbers))

# Check prime number
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True

# Prime numbers
prime_numbers = [x for x in numbers if is_prime(x)]

print("\nPrime numbers:")
print(prime_numbers)
print("Count of prime numbers:", len(prime_numbers))

# Most frequent number
frequency = Counter(numbers)

most_frequent_number, frequency_count = frequency.most_common(1)[0]

print("\nMost frequent number:", most_frequent_number)
print("Number of times it occurs:", frequency_count)
```

```python
# Q4. Sets

# Enter roll number
roll_no = input("Enter your roll number: ")

# Take the first 8 digits
digits = [int(digit) for digit in roll_no[:8]]

# Create sets A and B
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)

# Union
print("\nUnion:", A.union(B))

# Intersection
print("Intersection:", A.intersection(B))

# A - B
print("A - B:", A.difference(B))

# B - A
print("B - A:", B.difference(A))

# Symmetric difference
print("Symmetric Difference:", A.symmetric_difference(B))

# Subset and superset
print("\nIs A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

# Take value from user
X = int(input("\nEnter a value to remove from A: "))

# discard() removes X if present
A.discard(X)

print("Set A after discard:", A)

# discard() does not give an error if X is not present,
# unlike remove(), which raises an error.
```

```python
# Q5. Dictionaries

# Taking details as input
name = input("Enter your name: ")
roll_no = input("Enter your roll number: ")
branch = input("Enter your branch: ")
age = int(input("Enter your age: "))
city = input("Enter your home city: ")
cgpa = float(input("Enter your CGPA: "))

# Creating dictionary
my_dict = {
    "name": name,
    "roll_no": roll_no,
    "branch": branch,
    "age": age,
    "city": city
}

print("Original Dictionary:")
print(my_dict)

# i. Rename city to location
my_dict["location"] = my_dict.pop("city")

print("\nAfter renaming city to location:")
print(my_dict)

# ii. Add CGPA
my_dict["cgpa"] = cgpa

print("\nAfter adding CGPA:")
print(my_dict)

# iii. Increase age by 1
my_dict["age"] += 1

print("\nAfter increasing age by 1:")
print(my_dict)

# iv. Delete branch using pop()
dict_pop = my_dict.copy()
removed_value = dict_pop.pop("branch")

print("\nUsing pop():")
print(dict_pop)
print("Removed value:", removed_value)

# Delete branch using del
dict_del = my_dict.copy()
del dict_del["branch"]

print("\nUsing del:")
print(dict_del)

# pop() returns the deleted value, while del only deletes the key.

# v. Iterate using items()
print("\nKey-value pairs:")

for key, value in my_dict.items():
    print(key, "→", value)

# vi. Check whether email exists
print("\nChecking email:")

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key is not present.")

# vii. Create friend dictionary
friend_dict = {
    "name": "Rahul",
    "roll_no": "99999999",
    "branch": "CSE",
    "age": 20,
    "city": "Delhi"
}

print("\nFriend Dictionary:")
print(friend_dict)

# Merge dictionaries
merged_dict = {**my_dict, **friend_dict}

print("\nMerged Dictionary:")
print(merged_dict)

# When both dictionaries have the same key,
# the value from the second dictionary wins.

# viii. Dictionary comprehension
string_values_dict = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nDictionary with only string values:")
print(string_values_dict)
```
